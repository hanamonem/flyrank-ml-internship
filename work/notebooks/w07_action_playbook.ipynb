## 2. Intended use and limits

### Intended Use

This playbook is intended to help content teams prioritize refresh opportunities.

The output should be used as a decision-support ranking rather than an automated publishing system.

### Limits

The analysis is observational.

External events, seasonality, search engine updates, and content quality factors may affect outcomes.

The recommendations indicate directional opportunities and should not be interpreted as guaranteed performance improvements.

```python
print("Intended use and limitations documented.")
```

## 3. Human review + the no-go list
### Human Review Requirements

Every recommendation should be reviewed by a content editor before action.

Editors should verify:

- Content accuracy
- Business relevance
- Search intent alignment
- Editorial quality

### No-Go List

The following actions should NOT be automated:

- Publishing content updates automatically
- Deleting existing content automatically
- Rewriting content without human review
- Making business decisions solely from model outputs

The model is a prioritization aid, not a replacement for editorial judgment.

```python
print("Human review requirements documented.")
```

## 4. Monitoring / retrain triggers
### Monitoring

### Retrain Triggers

Retraining should be considered when:

- Search behavior changes substantially
- Feature distributions shift
- Ranking performance deteriorates

## 5. Exports for the paper
The notebook exports a ranked recommendation queue for potential use in the final paper.

The exported file is intended for demonstration and reproducibility purposes.

No client-identifying details are included in exported outputs.

```python
import pandas as pd
import os

url = "https://raw.githubusercontent.com/hanamonem/flyrank-ml-internship/main/data/raw/content_refresh_anonymized.csv"

df = pd.read_csv(url)

queue = df[
    [
        "search_volume",
        "impressions_last_30d",
        "clicks_last_30d",
        "ctr",
        "trend_direction",
        "trend_pct"
    ]
].copy()

queue = queue.sort_values(
    "trend_pct"
)

os.makedirs("work/outputs", exist_ok=True)

queue.head(50).to_csv(
    "work/outputs/action_queue.csv",
    index=False
)

print("Export created:")
print("work/outputs/action_queue.csv")
print(queue.head())
```
