## 1. My lane as an ML task (Ranking)
to order content or pages by their expected relevance and buisness value the model should rank the most useful results highwe than less useful ones


```python
import pandas as pd

url = "https://raw.githubusercontent.com/hanamonem/flyrank-ml-internship/refs/heads/main/data/raw/content_refresh_anonymized.csv"

df = pd.read_csv(url)

df.head()
```

```python
df[['ctr', 'avg_position', 'trend_pct']].describe()
```

## 2. Target or proxy
the target is content ranking priority . as a proxy i would use observed user and search engine outcomes rather than a manually defined rule

```python
df[['trend_pct', 'ctr', 'avg_position']].head()
```

## 3. Success metric
NDCG is an appropriate metric because it measures how well highly relevant items appear near the top of the ranking . a higgher NDCG score indicates better ranking quality

```python
df['trend_direction'].value_counts()
```

## 4. The unit of analysis, as a real dataframe

one raw represents one content page each row contains features describing the page , such as traffic engagement metrics content characteristics and other signals used to determine ranking quality

```python
print("Rows:", len(df))
print("Columns:", len(df.columns))

df.head()
```

## 5. Why ML beats a fixed rule here

ranking quality dependes on many factors that interact in complex ways A fixed rule such as sorting only by traffic would ignore important signals like engagement . content freshness . machine learning can learn these patterns from data and produce better rankings than simple if-statement

```python
df[['search_volume',
    'ctr',
    'avg_position',
    'engagement_rate',
    'trend_pct']].corr()
```
